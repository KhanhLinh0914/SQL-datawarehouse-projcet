```python
#Read cust_info CSV file with pandas
import pandas as pd 
cust_info_crm = pd.read_csv("/Workspace/Users/phamthikhanhlinh0914@gmail.com/cust_info.csv")
display(cust_info_crm)
```

```python
#Using Spark to convert dataframe to Table Bronze.crm_cust_info 
spark_cust_info = spark.createDataFrame(cust_info_crm)
spark_cust_info.write.format("delta").mode("overwrite").option("overwriteSchema", "true").saveAsTable("bronze.crm_cust_info")
```

```python
result = spark.sql("select * from bronze.crm_cust_info")
display(result)
```
